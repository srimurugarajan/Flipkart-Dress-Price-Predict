```python
# Step 1: Import Libraries and Setup Sample Flipkart Dress Dataset
import pandas as pd
import numpy as np

# Sample data creation for Flipkart Dress Price Prediction
data = {
    'Brand': ['Trooper', 'Badeyes', 'Shakethat', 'Bloodysweet', 'Powerhouse', 'Ordinaryperson', 'Preservance', 'Imscared','Anifield'],
    'Material': ['Cotton', 'Polyester', 'Cotton', 'synthetic', 'Cottonsilk', 'Silkworm', 'Cottonmix', 'woolentype','cotton'],
    'Rating': [4.9, 4.9, 4.5, 4.0, 4.3, 4.6, 3.8, 4.1,4.5],
    'Discount_Price (INR)': [6000, 1850, 600, 990, 750, 2500, 900, 950,1220]
}

# Convert to DataFrame
df = pd.DataFrame(data)

# Display the first few rows of our dataset
print("Dataset Successfully Loaded! Here are the first few rows:")
display(df)
```

```python
# Retrain model with new brand names
from sklearn.linear_model import LinearRegression

df_encoded = pd.get_dummies(df, columns=['Brand', 'Material'])
X = df_encoded.drop('Discount_Price (INR)', axis=1)
y = df_encoded['Discount_Price (INR)']

model = LinearRegression()
model.fit(X, y)

print("Model successfully updated with new brand names! 🚀")
```

```python
# Predict the price for the updated dress data
new_dress = X.tail(1) # last dress details
predicted_price = model.predict(new_dress)

print(f"✨ Predicted Price for the dress: ₹ {predicted_price[0]:.2f}")
```
